```python
import pandas as pd
import numpy as np
import os

base = '/workspaces/hospital-financial-risk-analysis/data'

files = {
    2019: f'{base}/Hospital_Provider_Cost_Report_2019.csv',
    2020: f'{base}/Hospital_Provider_Cost_Report_2020.csv',
    2021: f'{base}/Hospital_Provider_Cost_Report_2021.csv',
    2022: f'{base}/Hospital_Provider_Cost_Report_2022.csv',
    2023: f'{base}/Hospital_Provider_Cost_Report_2023_updated.csv',
}

dfs = []
for year, file in files.items():
    df = pd.read_csv(file)
    df['year'] = year
    dfs.append(df)
    print(f"Loaded {year}: {df.shape[0]} rows")

df = pd.concat(dfs, ignore_index=True)
print(f"\nCombined: {df.shape}")
```

```python
# See all columns clearly
for i, col in enumerate(df.columns.tolist()):
    print(f"{i}: {col}")
```

```python
print(df.columns[24:114].tolist())
```

```python
# Select only the columns we care about
cols_to_keep = [
    'Provider CCN', 'Hospital Name', 'City', 'State Code', 
    'Rural Versus Urban', 'Type of Control', 'Provider Type',
    'Number of Beds', 'FTE - Employees on Payroll',
    'Total Days Title XVIII', 'Total Days Title XIX',
    'Total Discharges (V + XVIII + XIX + Unknown)',
    'Cost of Charity Care', 'Total Bad Debt Expense',
    'Cost of Uncompensated Care', 'Total Unreimbursed and Uncompensated Care',
    'Total Costs', 'Total Patient Revenue', 'Net Patient Revenue',
    'Net Income', 'Net Revenue from Medicaid', 'Medicaid Charges',
    'Total Assets', 'Total Liabilities', 'Net Income from Service to Patients',
    'Disproportionate Share Adjustment', 'year'
]

df_clean = df[cols_to_keep].copy()
print(df_clean.shape)
print(df_clean.head())
```

```python
# Check missing values
print("Missing values per column:")
print(df_clean.isnull().sum())
print(f"\nTotal missing: {df_clean.isnull().sum().sum()}")

# Check data types
print("\nData types:")
print(df_clean.dtypes)

# Basic statistics
print("\nBasic statistics:")
print(df_clean.describe())
```

```python
# Drop rows where our target variable is missing
df_clean = df_clean.dropna(subset=['Cost of Uncompensated Care'])
print(f"Rows after dropping missing target: {df_clean.shape[0]}")

# Check remaining missing values
print("\nRemaining missing values:")
print(df_clean.isnull().sum())
```

```python
# Fill numerical columns with median (safer than mean for financial data)
numerical_cols = [
    'Number of Beds', 'FTE - Employees on Payroll',
    'Total Days Title XVIII', 'Total Days Title XIX',
    'Total Discharges (V + XVIII + XIX + Unknown)',
    'Cost of Charity Care', 'Total Bad Debt Expense',
    'Total Patient Revenue', 'Net Patient Revenue',
    'Net Income', 'Net Revenue from Medicaid', 'Medicaid Charges',
    'Total Assets', 'Total Liabilities', 'Net Income from Service to Patients',
    'Disproportionate Share Adjustment'
]

for col in numerical_cols:
    median_val = df_clean[col].median()
    df_clean[col] = df_clean[col].fillna(median_val)

# Total Days Title XIX has 2110 missing - fill with 0 
# (some hospitals may genuinely have no Medicaid patients)
df_clean['Total Days Title XIX'] = df_clean['Total Days Title XIX'].fillna(0)

# Verify no missing values remain
print("Missing values after cleaning:")
print(df_clean.isnull().sum())
print(f"\nFinal dataset shape: {df_clean.shape}")
```

```python
# Save cleaned dataset
df_clean.to_csv('../data/cleaned_hospital_data.csv', index=False)
print("Cleaned dataset saved successfully!")
print(f"Final shape: {df_clean.shape}")
```
